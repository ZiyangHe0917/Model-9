```python
def lbs_to_kg(pounds):
    """
    Converts weight from pounds to kilograms.
    Formula: 1 lb = 0.453592 kg
    """
    kilograms = pounds * 0.453592
    return kilograms

weight_1 = 150
print(f"{weight_1} lbs is equal to {lbs_to_kg(weight_1):.2f} kg")

weight_2 = 200
print(f"{weight_2} lbs is equal to {lbs_to_kg(weight_2):.2f} kg")

my_weight = 165
print(f"My weight ({my_weight} lbs) is equal to {lbs_to_kg(my_weight):.2f} kg")
```
